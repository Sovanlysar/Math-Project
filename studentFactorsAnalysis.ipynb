```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
df = pd.read_csv('data_sets/StudentPerformanceFactors.csv').dropna().drop_duplicates()
df.shape
```

```python
df.info()
```

```python
df.head()
df = df.drop(columns=["Gender"])
# Clean headers
df.columns = df.columns.str.strip()

# Clean string cell values across all applicable columns
df = df.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
```

```python
df.dtypes
```

```python
df.columns
```

```python
df.info()
```

```python
# binary categories: Extracurricular_Activities Internet_Access School_Type 
# Learning_Disabilities
cols = ["Extracurricular_Activities", "Learning_Disabilities", "School_Type", "Internet_Access"]
target_cols = [c for c in cols if c in df.columns]

if target_cols:
    df = pd.get_dummies(df, columns=target_cols, drop_first=True)

df.head()
```

```python
df.describe()
```

```python
df["Exam_Score"] = df["Exam_Score"].clip(lower=0, upper=100)
df.describe()
```

```python
# categories: low medium high
# Access_to_Resources, Motivation_Level, Family_Income, Teacher_Quality
# Peer_Influence, Parental_Education_Level
# Distance_from_Home
education_map = {"High School": 0, 
                "College": 1, 
                "Postgraduate": 2}

df["Parental_Education_Level"] = df["Parental_Education_Level"].map(education_map)
df.head()
```

```python
peer_influence_map = {"Negative": 0,
                      "Neutral": 1,
                      "Positive": 2}
df["Peer_Influence"] = (
    df["Peer_Influence"].astype(str).str.strip().map(peer_influence_map)
)
```

```python
low_medium_high_map = {"Low": 0,
                       "Medium": 1,
                       "High": 2}
cols = ["Parental_Involvement", "Access_to_Resources", "Motivation_Level",
        "Family_Income", "Teacher_Quality"]
df[cols] = df[cols].apply(lambda col: col.str.strip()).replace(low_medium_high_map)
```

```python
distance_home_map = {"Near": 0,
                     "Moderate": 1,
                     "Far": 2}
df["Distance_from_Home"] = (
    df["Distance_from_Home"].str.strip().replace(distance_home_map)
)
```

```python
X = df.drop(columns=["Exam_Score"])
X = sm.add_constant(X)
y = df.Exam_Score

model = sm.OLS(y, X.astype(float)).fit()
print(model.summary())
```
